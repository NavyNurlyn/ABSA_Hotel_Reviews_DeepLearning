```python
import pandas as pd

file_path = '/content/drive/MyDrive/SKRIPSI_NAVY/Dataset/Dataset_Persiapan/Dataset_Translated/hotel_reviews_translated.csv'
df = pd.read_csv(file_path)

print("Ukuran dataset:", df.shape)
df
```

```python
# Menentukan Kolom untuk anotasi/pelabelan (id, review_text_translated, review_text)
annotation_df = df[['id', 'review_text', 'review_text_translated']].copy()
annotation_df['Room'] = ''
annotation_df['Hotel'] = ''
annotation_df['Location'] = ''
annotation_df['Service'] = ''

annotation_df
```

```python
from openpyxl import load_workbook
from openpyxl.worksheet.datavalidation import DataValidation

base_path = '/content/drive/MyDrive/SKRIPSI_NAVY/Dataset/Dataset_Persiapan/Dataset_Pelabelan/RAW/Pelabelan_Annotator_1.xlsx'
second_path = '/content/drive/MyDrive/SKRIPSI_NAVY/Dataset/Dataset_Persiapan/Dataset_Pelabelan/Pelabelan_Annotator_1.xlsx'

# Simpan file
annotation_df.to_excel(base_path, index=False)

# Load kembali untuk tambah dropdown
wb = load_workbook(base_path)
ws = wb.active

# Buat dropdown validation
dv = DataValidation(
    type="list",
    formula1='"None,Positive,Negative"',
    allow_blank=True
)

# Tambahkan dropdown ke worksheet
ws.add_data_validation(dv)

# Tentukan range kolom aspek (kolom D sampai G)
max_row = ws.max_row
dv.add(f'D2:D{max_row}')  # Aspek Room
dv.add(f'E2:E{max_row}')  # Aspek Hotel
dv.add(f'F2:F{max_row}')  # Aspek Location
dv.add(f'G2:G{max_row}')  # Aspek Service

# Simpan ke lokasi pertama
wb.save(base_path)
print(f"File annotator_1.xlsx berhasil dibuat di {base_path}!")

# Simpan ke lokasi kedua
wb.save(second_path)
print(f"File annotator_1.xlsx berhasil disimpan juga di {second_path}!")
```

```python
# File pelabelan untuk anotator 2 dan 3
for i in range(2, 4):
    raw_path = f'/content/drive/MyDrive/SKRIPSI_NAVY/Dataset/Dataset_Persiapan/Dataset_Pelabelan/RAW/Pelabelan_Annotator_{i}.xlsx'
    general_path = f'/content/drive/MyDrive/SKRIPSI_NAVY/Dataset/Dataset_Persiapan/Dataset_Pelabelan/Pelabelan_Annotator_{i}.xlsx'

    # Simpan file
    annotation_df.to_excel(raw_path, index=False)

    # Load kembali untuk menambahkan dropdown
    wb = load_workbook(raw_path)
    ws = wb.active

    dv = DataValidation(
        type="list",
        formula1='"None,Positive,Negative"',
        allow_blank=True
    )

    ws.add_data_validation(dv)

    max_row = ws.max_row
    dv.add(f'D2:D{max_row}')
    dv.add(f'E2:E{max_row}')
    dv.add(f'F2:F{max_row}')
    dv.add(f'G2:G{max_row}')

    # Simpan kembali (setelah penambahan dropdown)
    wb.save(raw_path)
    print(f"File annotator_{i}.xlsx berhasil dibuat di {raw_path}!")

    # Simpan ke path lain
    wb.save(general_path)
    print(f"File annotator_{i}.xlsx berhasil disimpan juga di {general_path}!")

print("Semua file anotator berhasil dibuat!")
```
