## Translate Library GoogleTrans

```python
!pip install googletrans==4.0.0-rc1
!pip install langdetect
import pandas as pd
from langdetect import detect, LangDetectException
from googletrans import Translator
from tqdm import tqdm

tqdm.pandas()
```

```python
def detect_language(text):
    try:
        return detect(str(text))
    except LangDetectException:
        return "unknown"

df_translate['detected_language'] = df_translate['review_text'].progress_apply(detect_language)
translator = Translator()
def translate_to_indonesian(text, lang):
    try:
        if lang != 'id' and lang != 'unknown':
            return translator.translate(text, src=lang, dest='id').text
        else:
            return text
    except:
        return text

df_translate['review_text_translated'] = df_translate.progress_apply(
    lambda row: translate_to_indonesian(row['review_text'], row['detected_language']),
    axis=1
)
```
